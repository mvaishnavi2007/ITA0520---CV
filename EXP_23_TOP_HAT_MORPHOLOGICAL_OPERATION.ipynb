```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]
img = cv2.imdecode(
    np.frombuffer(uploaded[filename], np.uint8),
    cv2.IMREAD_COLOR
)

if img is None:
    raise ValueError("Upload a valid image.")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (15, 15)
)

print("Kernel created successfully")
```

```python
tophat = cv2.morphologyEx(
    gray,
    cv2.MORPH_TOPHAT,
    kernel
)

print("Top Hat operation completed")
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(tophat, cmap="gray")
plt.title("Top Hat Image")
plt.axis("off")

plt.show()
```

```python
cv2.imwrite("tophat_image.jpg", tophat)
files.download("tophat_image.jpg")
```
