```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[filename], np.uint8),
    cv2.IMREAD_COLOR
)

if img is None:
    raise ValueError("Image not found. Upload a JPG or PNG image.")

print("Image loaded successfully!")
print("Image dimensions:", img.shape)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
rotated = cv2.rotate(img, cv2.ROTATE_180)

print("Image rotated by 180 degrees successfully!")
```

```python
rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

plt.imshow(rotated_rgb)
plt.title("Image Rotated by 180 Degrees")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rotated_rgb)
plt.title("180 Degree Rotated Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite("rotated_180_image.jpg", rotated)

print("Output image saved successfully!")
```
