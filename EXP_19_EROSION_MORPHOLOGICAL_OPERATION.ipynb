```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[filename], np.uint8),
    cv2.IMREAD_COLOR
)

if img is None:
    raise ValueError("Please upload a valid image.")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.title("Original Gray-scale Image")
plt.axis("off")
plt.show()
```

```python
ret, binary = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")
plt.show()
```

```python
kernel = np.ones((5, 5), np.uint8)

print("Kernel created successfully")
print(kernel)
```

```python
erosion = cv2.erode(
    binary,
    kernel,
    iterations=1
)

print("Morphological erosion completed")
```

```python
plt.imshow(erosion, cmap="gray")
plt.title("Erosion Morphological Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(binary, cmap="gray")
plt.title("Original Binary Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(erosion, cmap="gray")
plt.title("Eroded Image")
plt.axis("off")

plt.show()
```

```python
cv2.imwrite("morphological_erosion.jpg", erosion)
files.download("morphological_erosion.jpg")
```
