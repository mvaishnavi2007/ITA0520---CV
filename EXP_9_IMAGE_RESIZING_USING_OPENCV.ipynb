```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[filename], dtype=np.uint8),
    cv2.IMREAD_COLOR
)

if img is None:
    raise ValueError("Image could not be read. Upload the downloaded JPG or PNG image.")

print("Image loaded successfully!")
print("Image dimensions:", img.shape)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(img_rgb)
plt.title("Original Input Image")
plt.axis("off")
plt.show()
```

```python
resized = cv2.resize(img, (300, 300))

print("Image resized successfully!")
print("Resized dimensions:", resized.shape)
```

```python
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(5, 5))
plt.imshow(resized_rgb)
plt.title("Resized Image - 300 x 300")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(resized_rgb)
plt.title("Resized Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite("resized_image.jpg", resized)

print("Resized image saved successfully!")

files.download("resized_image.jpg")
```
