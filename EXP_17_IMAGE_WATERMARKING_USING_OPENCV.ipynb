```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

img = cv2.imdecode(
    np.frombuffer(uploaded[filename], np.uint8),
    cv2.IMREAD_COLOR
)

if img is None:
    raise ValueError("Please upload a valid image.")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
watermarked = img.copy()

height, width = img.shape[:2]

text = "MY WATERMARK"

position = (20, height - 30)

font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
color = (255, 255, 255)
thickness = 2

cv2.putText(
    watermarked,
    text,
    position,
    font,
    font_scale,
    color,
    thickness,
    cv2.LINE_AA
)

print("Watermark added successfully")
```

```python
watermarked_rgb = cv2.cvtColor(
    watermarked, cv2.COLOR_BGR2RGB
)

plt.imshow(watermarked_rgb)
plt.title("Watermarked Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(watermarked_rgb)
plt.title("Watermarked Image")
plt.axis("off")

plt.show()
```

```python
cv2.imwrite("watermarked_image.jpg", watermarked)
files.download("watermarked_image.jpg")
```
